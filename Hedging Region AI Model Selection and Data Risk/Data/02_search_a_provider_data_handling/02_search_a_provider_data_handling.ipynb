## 2. Table structure and coding scheme

### 2.1 Output structure: `processed/provider_data_handling.csv`

One row per provider and tier (12 rows).

| Column | Type | Allowed values or example |
|---|---|---|
| `provider` | text | OpenAI, Anthropic, Google, DeepSeek, Alibaba Cloud, Z.ai (Zhipu), Zhipu, Moonshot AI (Kimi) |
| `tier` | text | api, consumer, paid, unpaid, app_web, platform |
| `site` | text | international or china |
| `training_default` | text | no_by_default, yes_by_default, depends_on_tier, contract_only, not_stated |
| `quote` | text | exact words from the page that support `training_default`, empty if we paraphrase |
| `quote_language` | text | en, or zh_translated when the page is Chinese and the quote is our translation |
| `retention_class` | text | fixed_days, account_lifetime, not_retained, not_stated |
| `retention_days` | integer or empty | 30 |
| `zero_retention_option` | text | yes_on_request, no, not_stated |
| `storage_location` | text | a stated country or region, or not_stated |
| `region_choice` | text | yes, no, not_stated |
| `authority_sharing` | text | clause_present, not_stated, not_checked |
| `contracting_entity` | text | company that signs the terms, or not_checked |
| `governing_law` | text | law named by the terms, or not_stated, or not_checked |
| `policy_date` | date or empty | effective or last-updated date printed on the page |
| `date_viewed` | date | 2026-10-16 |
| `source_url` | text | main page the row was read from |
| `other_source_urls` | text | further pages, joined with a vertical bar |
| `epoch_organization` | text | the matching `organization` value in notebook 01 (Epoch's ECI table), for the join. Empty if Epoch has no such developer name |
| `note` | text | caveats from the project note |

### 2.2 Coding rules (from the Search A brief)

- `training_default`: code from the provider's own statement about using customer inputs and outputs to train or improve models. If the page is silent, code `not_stated`. Do not infer it from reputation. `yes_by_default` means training is allowed by default with no contract route stated. `contract_only` means training is allowed by default and only a written agreement can restrict it.
- `retention_class`: `fixed_days` needs a number of days on the page. `account_lifetime` means kept while the account is active. `not_retained` means the page says inputs are not saved.
- `zero_retention_option`: `yes_on_request` means the page says a zero data retention arrangement exists, even if approval is needed.
- `storage_location`: record the place exactly as the page words it. If the page says only where the service is available, code `not_stated`.
- `region_choice`: `yes` means the customer can pick a storage or processing region.
- `authority_sharing`: `clause_present` means the page says data may be shared with public authorities. This records that a clause exists and not how often it is used.
- `policy_date`: the effective or last-updated date printed on the page, otherwise empty. Where a row rests on two pages with different dates, the date of the page that holds the training statement is used and the other is in `note`.

## 5. Build the table and validate it

The checks follow the three tests in the Search A brief: every row has a source link and a viewing date, no value falls outside the allowed list, and provider names match the model table so that a join drops no rows.

```python
ALLOWED = {
    "training_default": {"no_by_default", "yes_by_default", "depends_on_tier", "contract_only", "not_stated"},
    "retention_class": {"fixed_days", "account_lifetime", "not_retained", "not_stated"},
    "zero_retention_option": {"yes_on_request", "no", "not_stated"},
    "region_choice": {"yes", "no", "not_stated"},
    "authority_sharing": {"clause_present", "not_stated", "not_checked"},
    "site": {"international", "china"},
}
COLUMNS = ["provider", "tier", "site", "training_default", "quote", "quote_language", "retention_class",
           "retention_days", "zero_retention_option", "storage_location", "region_choice", "authority_sharing",
           "contracting_entity", "governing_law", "policy_date", "date_viewed", "source_url",
           "other_source_urls", "epoch_organization", "note"]

table = pd.DataFrame(RECORDS)[COLUMNS]
table["policy_date"] = pd.to_datetime(table["policy_date"], errors="coerce")
table["date_viewed"] = pd.to_datetime(table["date_viewed"])
table["retention_days"] = pd.array(table["retention_days"], dtype="Int64")

problems = []
for column, allowed in ALLOWED.items():
    bad = table.loc[~table[column].isin(allowed), column]
    if len(bad):
        problems.append(f"{column}: values outside the list {sorted(bad.unique())}")
if table["source_url"].isna().any() or not table["source_url"].str.startswith("http").all():
    problems.append("a row has no valid source_url")
if table["date_viewed"].isna().any():
    problems.append("a row has no date_viewed")
if table.duplicated(subset=["provider", "tier", "site"]).any():
    problems.append("duplicated provider, tier and site")
fixed = table[table["retention_class"].eq("fixed_days")]
if fixed["retention_days"].isna().any():
    problems.append("fixed_days without a number of days")
if table.loc[~table["retention_class"].eq("fixed_days"), "retention_days"].notna().any():
    problems.append("retention_days given for a row that is not fixed_days")

# Join test against notebook 01 (Epoch). Skipped with a message if notebook 01 has not been run yet.
epoch_file = DATA / "01_epoch_eci_scores" / "processed" / "model_capability.csv"
if epoch_file.exists():
    epoch_orgs = set(pd.read_csv(epoch_file)["organization_primary"].dropna())
    wanted = set(table["epoch_organization"].dropna())
    lost = sorted(wanted - epoch_orgs)
    print("Organisation names in this table that Epoch does not use:", lost if lost else "none")
    if lost:
        problems.append(f"join would drop {lost}")
else:
    print("Notebook 01 has not been run yet, so the join test was skipped.")

print("Problems:", problems if problems else "none")
assert not problems, problems
table.head(3)
```

## 6. Optional: check that the source links still open

This cell is **off by default** because it sends about 25 requests to provider websites and some of them block automated requests even when the page is fine in a browser. A failure here is a reminder to open the page by hand and not proof that the page is gone. Set `CHECK_LINKS = True` to run it. Expect 20 to 60 seconds.

```python
CHECK_LINKS = False

if CHECK_LINKS:
    import requests
    urls = sorted(set(table["source_url"]) | {u for cell in table["other_source_urls"].dropna() for u in cell.split("|")})
    results = []
    for url in urls:
        try:
            status = requests.get(url, headers={"User-Agent": "UTS-36104-A3-student-project (academic use)"},
                                  timeout=20, allow_redirects=True).status_code
        except Exception as err:
            status = type(err).__name__
        results.append({"url": url, "status": status})
        time.sleep(0.5)
    display(pd.DataFrame(results))
else:
    print("Link check skipped (CHECK_LINKS is False).")
```

## How to update

There is no file to download. The table is only as current as the day the pages were read (2026-10-16). To bring it up to date:

1. Open each link in Section 1 in a browser. No login is needed.
2. For each row in the cell `RECORDS`, compare the page with these fields: training default, retention, zero retention, storage location, region choice, authority sharing, contracting entity, governing law and the policy date printed on the page. Use the coding rules in Section 2.2.
3. Edit the row in `RECORDS` where the page changed. Change `policy_date` to the new printed date. Keep the old wording of anything that did not change.
4. Change `VIEWED` at the top of the same cell to the day you read the pages.
5. Run all cells. Section 5 stops with a message if a value is outside the allowed list, a link is missing, or a provider name no longer matches Epoch's table.
6. Optional: set `CHECK_LINKS = True` in Section 6 to test that every link still opens. Some sites block automated requests even when the page works in a browser, so a failure there means open the page by hand.
7. To add a provider (for example MiniMax, which is missing), add one `row(...)` entry with its own pages and quotes.

Remember that the pages were read through a tool that returns extracts, so re-read each quote on the live page before it goes into a graded document.

**Update check.** Open `processed/data_currency.csv`. `data_covers_to` should be your new viewing day.

## 9. Run time and limits

**Time to extract.** Nothing is downloaded, so the run takes **under 2 seconds**. The optional link check adds 20 to 60 seconds. The real cost of this dataset is the human reading time, not the code.

**Limits and known risks.**

1. The table records what providers say, not what they do.
2. Quotes came through a page reader that returns extracts. Re-read each quote on the live page before it goes into a graded document.
3. Policies change. The `policy_date` and `date_viewed` columns exist so the app can show how fresh each value is. Re-check the pages before submitting.
4. Singapore storage is a storage fact. It does not mean the data stays out of China, because Z.ai and Moonshot also run China sites under mainland law.
5. `authority_sharing` records that a clause exists. It does not measure how often it is used.
6. MiniMax is **not** in this table. If Q1 needs MiniMax, a new row must be read from MiniMax's own pages.
7. Tiers differ. Keep each row's `tier` visible in the app.

## Data currency: how up to date is this data?

Every notebook writes `processed/data_currency.csv` with the day it was extracted, the dates the data itself covers, and what we know about when the source started and how often it changes. Use these values for the "data as of" label in the app.

```python
def write_currency(dataset, covers_from, covers_to, snapshot_note, origin_note, frequency):
    """Record how current the data is. extracted_utc is the moment this notebook ran."""
    row = {
        "dataset": dataset,
        "extracted_utc": datetime.now(timezone.utc).strftime("%Y-%m-%dT%H:%M:%SZ"),
        "data_covers_from": str(covers_from)[:10] if covers_from is not None else "not stated in the data",
        "data_covers_to": str(covers_to)[:10] if covers_to is not None else "not stated in the data",
        "snapshot_note": snapshot_note,
        "source_origin_note": origin_note,
        "source_update_frequency": frequency,
    }
    pd.DataFrame([row]).to_csv(OUT / "data_currency.csv", index=False, encoding="utf-8")
    for key, value in row.items():
        print(f"{key:24s} {value}")


pd_dates = table["policy_date"].dropna()
write_currency("Search A provider data handling (provider_data_handling.csv)",
               pd_dates.min(), table["date_viewed"].max(),
               "hand-built table, pages read on 2026-10-16. data_covers_from and data_covers_to show the oldest policy date (2024 for the DeepSeek app policy) and the viewing day. The running of this notebook adds nothing new",
               "Each row's policy_date is the date printed on the provider page",
               "None. Re-read the pages before submission, policies change")
```
